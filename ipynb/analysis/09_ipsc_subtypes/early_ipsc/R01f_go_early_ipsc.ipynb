```python
import pandas as pd
from gprofiler import GProfiler
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2'
outdir = '/u/project/cluo/terencew/claude/igvf/pgp/ipsc_le_se_rerun/csv'
donors = ['C29', 'C37', 'C38', 'C39']
```

### ED10J: GO terms (padj < 0.05, both directions; 02b_main_alt_degs_gene_ontology used 1e-3), plus SE-up only

```python
alt = pd.read_csv(f'{outdir}/degs_ips_alt.csv', sep='\t', index_col=0)
thresh = 0.05
sig = alt[alt['pvals_adj'] < thresh]
gp = GProfiler(return_dataframe=True)
go_all = gp.profile(organism='hsapiens', query=list(sig['names']), sources=['GO:BP', 'GO:MF', 'GO:CC'], no_evidences=False)
go_all.shape
```

```python
go_all[['source', 'name', 'p_value', 'intersection_size', 'query_size']].head(15)
```

```python
up = list(alt[(alt['pvals_adj'] < thresh) & (alt['logfoldchanges'] > 0)]['names'])
go_up = gp.profile(organism='hsapiens', query=up, sources=['GO:BP', 'GO:MF', 'GO:CC'], no_evidences=False)
go_up[['source', 'name', 'p_value', 'intersection_size', 'query_size']].head(15)
```

```python
go_all.to_csv(f'{outdir}/ips_subtypes_go.csv', sep='\t')
go_up.to_csv(f'{outdir}/ips_subtypes_go_se_up.csv', sep='\t')
```
