```python
import numpy as np
import pandas as pd
import shutil
```

```python
projdir = '/u/project/cluo/terencew/claude/igvf/pgp/ipsc_le_se_rerun'
rdir = f'{projdir}/csv'
pubdir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure5'
outdir = pubdir
sfx = '_early_ipsc'
donors = ['C29', 'C37', 'C38', 'C39']
```

### straight copies (rerun name -> figure5 name + _early_ipsc)

```python
copies = {
    'ips_subtypes_dist_df': 'ips_subtypes_dist_df',
    'ips_subtypes_dist_meta': 'ips_subtypes_dist_meta',
    'ips_subtype_mcg_mature_corr_min10x': 'ips_subtype_mcg_mature_corr',
    'ips_subtype_mcg_umap_df': 'ips_subtype_mcg_umap_df',
    'ips_subtype_nn_prop': 'ips_subtype_nn_prop',
    'ips_subtype_nn_prop_balanced_500': 'ips_subtype_nn_prop_balanced_500',
    'ips_subtype_nn_prop_balanced_500_summary': 'ips_subtype_nn_prop_balanced_500_summary',
    'snmct_rna_ips_subtype_umap': 'snmct_rna_ips_subtype_umap',
    'ips_subtype_rna_traj_score': 'ips_subtype_rna_traj_score',
    'ips_subtype_rna_cell_cycle': 'ips_subtype_rna_cell_cycle',
    'ips_subtypes_pluri_markers': 'ips_subtypes_pluri_markers',
    'degs_ips_alt': 'ips_subtypes_degs_ips_alt',
    'ips_subtypes_go': 'ips_subtypes_go',
    'ips_cluster_donor_pmds_metagene': 'ips_cluster_donor_pmds_metagene',
    '4dn_mature_ips_eigenvector_corr': '4dn_mature_ips_eigenvector_corr',
    'ips_subtypes_tad_dist': 'ips_subtypes_tad_dist',
    'ips_subtypes_tads_great': 'ips_subtypes_tads_great',
    'ips_subtype_dmr_tad_boundary_overlap': 'ips_subtype_dmr_tad_boundary_overlap',
}
for src, dst in copies.items():
    shutil.copyfile(f'{rdir}/{src}.csv', f'{outdir}/{dst}{sfx}.csv')
len(copies)
```

### ED10B: LE/SE fractions per donor and per day (as in 20_check_ips_subtype_proportion_between_donors)

```python
early = pd.read_csv(f'{rdir}/ipsc_le_se_early_ipsc_nuclei.csv', sep='\t', index_col=0)
frac = pd.DataFrame([early[early['line'] == d]['ips_subtype'].value_counts(normalize=True) for d in donors], index=donors)[['IPS_main', 'IPS_alt']]
frac.to_csv(f'{outdir}/se_le_donor_prop{sfx}.csv', sep='\t')
frac
```

```python
tfrac = pd.DataFrame([early[early['ips_subtype'] == s]['time'].value_counts(normalize=True) for s in ['IPS_main', 'IPS_alt']], index=['IPS_main', 'IPS_alt']).T.sort_index().fillna(0)
tfrac.index = tfrac.index.astype(int)
tfrac.to_csv(f'{outdir}/se_le_time_prop{sfx}.csv', sep='\t')
tfrac
```

### Fig 6D / ED10C: labels for the 3C UMAP, early-iPSC nuclei only

```python
early[['ips_subtype', 'assay', 'line', 'time']].to_csv(f'{outdir}/snm3c_ips_subtype{sfx}.csv', sep='\t')
```

### Fig 6G / ED10D: published methylation UMAP coordinates, LE/SE labels restricted to early-iPSC nuclei

```python
for name in ['snmct_snm3c_mcg_ips_subtype_umap', 'snmct_snm3c_mcg_ips_subtype_umap_per_donor']:
    df = pd.read_csv(f'{pubdir}/{name}.csv', sep='\t', index_col=0)
    df['ips_subtype'] = early['ips_subtype'].reindex(df.index)
    df.to_csv(f'{outdir}/{name}{sfx}.csv', sep='\t')
    print(name, df.shape, df['ips_subtype'].value_counts().to_dict())
```
