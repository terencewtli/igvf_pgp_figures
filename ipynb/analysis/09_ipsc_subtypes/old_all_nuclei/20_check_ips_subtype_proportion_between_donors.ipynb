```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/mc'
donors = ['C29', 'C37', 'C38', 'C39']
```

```python
ips_clusters_path = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/mc/csv/label_transfer/snm3c_ips_subtype.csv'
ips_clusters = pd.read_csv(ips_clusters_path, sep='\t', index_col=0)
ips_clusters.shape
```

```python
ips_clusters.head()
```

```python
tmp = []
for d in donors:
    tmp_clusters = ips_clusters[ips_clusters['line'] == d]
    ips_counts = tmp_clusters['ips_subtype'].value_counts() 
    ips_frac = ips_counts / ips_counts.sum()
    tmp.append(ips_frac)
    
frac_df = pd.DataFrame(tmp, index=donors)
frac_df
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
frac_df.plot(kind='bar', stacked=True, ax=ax, color=['#3B4CC0', '#B40426'])
ax.set_xlabel('')
ax.set_ylabel('Proportion')
ax.legend(title='', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
outdir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure5'
frac_df.to_csv(f'{outdir}/se_le_donor_prop.csv', sep='\t')
```

### time proportion

```python
ips_order = ['IPS_main', 'IPS_alt']
tmp = []
for d in ips_order:
    tmp_clusters = ips_clusters[ips_clusters['ips_subtype'] == d]
    ips_counts = tmp_clusters['time'].value_counts() 
    ips_frac = ips_counts / ips_counts.sum()
    tmp.append(ips_frac)
    
frac_df = pd.DataFrame(tmp, index=ips_order).T.sort_index().fillna(0)
frac_df.index = frac_df.index.astype(int)
frac_df
```

```python
outdir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure5'
frac_df.to_csv(f'{outdir}/se_le_time_prop.csv', sep='\t')
```
