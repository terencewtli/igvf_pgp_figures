```python
hyper_path = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/pseudobulk_allc/csv/dmrs/ips_cluster/by_group/IPS_alt_hyper_dms1.csv'
hyper = pd.read_csv(hyper_path, sep='\t', header=None)
hyper.shape
```

```python
diff_path = '/u/project/cluo/terencew/claude/igvf/pgp/ips_subtypes/csv/enrich/ips_cluster_diff_bins.csv'
diff = pd.read_csv(diff_path, sep='\t', index_col=0)
diff.shape
```

```python
diff.head()
```

```python
nondiff_path = '/u/project/cluo/terencew/claude/igvf/pgp/ips_subtypes/csv/enrich/ips_cluster_nondiff_bins.csv'
nondiff = pd.read_csv(nondiff_path, sep='\t', index_col=0)
nondiff.shape
```

```python
nondiff.head()
```

```python
hypo
```

```python
hypo_bt = BedTool.from_dataframe(hypo)
hyper_bt = BedTool.from_dataframe(hyper)
diff_bt = BedTool.from_dataframe(diff)
nondiff_bt = BedTool.from_dataframe(nondiff)
```

```python
hypo_diff = hypo_bt.intersect(diff_bt, u=True).to_dataframe()
hyper_diff = hyper_bt.intersect(diff_bt, u=True).to_dataframe()
hypo_nondiff = hypo_bt.intersect(nondiff_bt, u=True).to_dataframe()
hyper_nondiff = hyper_bt.intersect(nondiff_bt, u=True).to_dataframe()
```

```python
hypo_diff.shape[0], hyper_diff.shape[0], hypo_nondiff.shape[0], hyper_nondiff.shape[0]
```

```python
hypo_diff.shape[0] / hypo.shape[0], hyper_diff.shape[0] / hyper.shape[0], \
hypo_nondiff.shape[0] / hypo.shape[0], hyper_nondiff.shape[0] / hyper.shape[0]
```

```python
to_plot = pd.DataFrame()
to_plot['overlap'] = [hypo_diff.shape[0] / hypo.shape[0], hyper_diff.shape[0] / hyper.shape[0],\
            hypo_nondiff.shape[0] / hypo.shape[0], hyper_nondiff.shape[0] / hyper.shape[0]]

dmr_types = ['hypo', 'hyper']
bin_types = ['diff', 'non-diff']
to_plot['dmr_type'] = np.tile(dmr_types, 2)
to_plot['bin_type'] = np.repeat(bin_types, 2)
to_plot
```

```python
sns.barplot(to_plot, x='dmr_type', y='overlap', hue='bin_type')
```

```python
outdir = "/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure5"
to_plot.to_csv(f'{outdir}/ips_subtype_dmr_tad_boundary_overlap.csv', sep='\t')
```
