### look at which bins are in the tail

```python
mask = boundaries['IPS_alt_frac'] >= boundaries['IPS_alt_frac'].quantile(0.9)
tmp_boundaries = boundaries[mask].sort_values(by='IPS_alt_frac', ascending=False)
tmp_boundaries.shape
```

```python
tmp_boundaries.head()
```

```python
sorted_boundaries = tmp_boundaries.copy()
sorted_boundaries['chr_num'] = [int(x.split('_')[0].split('chr')[1]) for x in sorted_boundaries.index]
sorted_boundaries['pos'] = [int(x.split('_')[1]) for x in sorted_boundaries.index]
sorted_boundaries = sorted_boundaries.sort_values(by=['chr_num', 'pos']).iloc[:,:-2]
sorted_boundaries.head()
```

```python
s = 'chr16_3025000'
x = np.where(sorted_boundaries.index == s)[0][0]
y = x + 50
sorted_boundaries.iloc[x:y]
```
