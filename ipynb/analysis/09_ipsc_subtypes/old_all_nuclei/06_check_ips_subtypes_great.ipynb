```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from concurrent.futures import ProcessPoolExecutor
from tqdm import tqdm
```

```python
great_path = '/u/project/cluo/terencew/igvf/2023_YR2/snm3C/hicluster/csv/great/ips_alt.csv'
great = pd.read_csv(great_path, sep='\t')
great.shape
```

```python
great['neg_log10p'] = -1 * np.log(great['binom_p_value'])
```

```python
sns.histplot(great['binom_p_value'])
```
