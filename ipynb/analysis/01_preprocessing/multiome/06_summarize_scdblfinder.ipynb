### summarize total doublet numbers

```python
dbl_summary = pd.DataFrame(index=samples, columns=['singlet', 'doublet'])

s = samples[0]
indir = f'{projdir}/doublets/dblfinder/gex/ambimux/wgs/{s}/'
summary = pd.read_csv(f'{indir}/scDblFinder_doublet_summary.tsv', sep='\t', header=0, index_col=0)
dbl_summary.loc[s] = summary.transpose().values

for s in samples[1:]:
    print(s)
    indir = f'{projdir}/doublets/dblfinder/gex/ambimux/wgs/{s}/'
    summary = pd.read_csv(f'{indir}/scDblFinder_doublet_summary.tsv', sep='\t', header=0, index_col=0)
    dbl_summary.loc[s] = summary.transpose().values

denom = np.sum(dbl_summary, axis=1)
dbl_summary['total_droplets'] = denom
dbl_summary['doublet_rate'] = dbl_summary['doublet'] / denom
```

```python
s = samples[0]
indir = f'{projdir}/doublets/dblfinder/gex/ambimux/wgs/{s}/'
dbl_scores = pd.read_csv(f'{indir}/scDblFinder_doublets_singlets.tsv', sep='\t', header=0, index_col=0)
dbl_scores['sample'] = s

for s in samples[1:]:
    print(s)
    indir = f'{projdir}/doublets/dblfinder/gex/ambimux/wgs/{s}/'
    tmp_scores = pd.read_csv(f'{indir}/scDblFinder_doublets_singlets.tsv', sep='\t', header=0, index_col=0)
    tmp_scores['sample'] = s
    dbl_scores = pd.concat([dbl_scores, tmp_scores])
    
dbl_scores['barcode'] = dbl_scores.index
dbl_scores.index = [f'{x}_{y}' for x,y in zip(dbl_scores['barcode'], dbl_scores['sample'])]

dbl_scores.shape
```
