```python
explore_cols =  ['frac_readnames_yieldcontact',
                 'frac_contacts_passfilt',
                 'frac_finalcont_inter',
                 'frac_finalcont_intraless1kb_totintra',
                 'frac_finalcont_intragreater1kb_totintra',
                 'frac_finalcont_intragreater10kb_totintra']
for col in explore_cols:
    print(meta[col].head()) 
```

```python
fig, axes = plt.subplots(3, 2, figsize=(20, 10))
for i, ax in enumerate(axes.flatten()):
    if i < len(explore_cols):
        qc_col = explore_cols[i]
        sns.histplot(meta, x=qc_col, ax=ax)
        vmax = sorted(meta[qc_col], reverse=True)[10]
        if pd.isna(vmax):
            vmax = 0
        print(vmax)
        ax.set_xlim(0, vmax)
        ax.set_title(qc_col)
        ax.set_xlabel('')
        ax.set_ylabel('')

plt.tight_layout()
```

```python
plot_cols = ['MappingRate_Total', 'UniqueMappedReads_PostSplit',
             'contacts_final_intra', 'frac_finalcont_intragreater1kb_totintra']

fig, axes = plt.subplots(2, 2, figsize=(10, 10))
for i, ax in enumerate(axes.flatten()):
    if i < len(plot_cols):
        qc_col = plot_cols[i]
        sns.histplot(meta, x=qc_col, ax=ax)
        vmax = sorted(meta[qc_col], reverse=True)[10]
        if pd.isna(vmax):
            vmax = 0
        print(vmax)
        ax.set_xlim(0, vmax)
        ax.set_title(qc_col)
        ax.set_xlabel('')
        ax.set_ylabel('')
        ax.set_xticks(ax.get_xticks(), fontsize=15)
        ax.set_yticks(ax.get_yticks(), fontsize=15)
        ax.set_title(qc_col, fontsize=20)

plt.tight_layout()
```
