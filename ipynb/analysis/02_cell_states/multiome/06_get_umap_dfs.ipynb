```python
clusters_path = '/u/project/cluo/terencew/igvf/2023_YR2/multiome/csv/clusters/rna_leiden_v7.csv'
clusters = pd.read_csv(clusters_path, sep='\t', index_col=0)
clusters.shape
```

```python
clusters.head()
```

```python
traj_score_path = '/u/project/cluo/terencew/igvf/2023_YR2/multiome/csv/atac/filtered_traj_score.csv'
traj_score = pd.read_csv(traj_score_path, sep='\t', index_col=0)
traj_score.index = [f'{x.split("#")[1]}_{x.split("#")[0]}' for x in traj_score.index]
traj_score.shape
```

```python
traj_score.head()
```

```python
adata.obs['coarse'] = clusters.reindex(adata.obs.index)['coarse_v7']
adata.obs['cluster'] = clusters.reindex(adata.obs.index)['clusters_v7']
adata.obs['time_num'] = clusters.reindex(adata.obs.index)['time_num']
adata.obs['traj_score'] = traj_score.reindex(adata.obs.index)['traj_score']
```

```python
umap_df = pd.DataFrame(index=adata.obs.index)
umap_df['umap0'] = adata.obsm['X_umap'][:,0]
umap_df['umap1'] = adata.obsm['X_umap'][:,1]

select = ['time_num', 'traj_score', 'cluster', 'coarse']
for s in select:
    umap_df[s] = adata.obs[s]
```

```python
scaler = MinMaxScaler()
umap_df['norm_traj'] = np.ravel(scaler.fit_transform(umap_df['traj_score'].values.reshape(-1,1)))
```

```python
indir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure1'
umap_df.to_csv(f'{indir}/multiome_atac_umap.csv', sep='\t')
```
