```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import scanpy as sc
import anndata
from sklearn.preprocessing import MinMaxScaler
```

```python
sc.set_figure_params(figsize=(5,5), frameon=False)
sc.settings.verbosity = 3
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/multiome'
```

```python
adata = sc.read_h5ad(f'{projdir}/h5ad/rna/processed/intersect_atac_reindex_small.h5ad')
```

```python
adata
```

```python
clusters = pd.read_csv(f'{projdir}/csv/clusters/rna_leiden_v7.csv', sep='\t', header=0, index_col=0)
adata.obs = clusters.reindex(adata.obs.index)
clusters.shape
```

```python
sc.pl.umap(adata, color='coarse_v7', title='')
```

```python
sc.pl.umap(adata, color='time_num', title='')
```

```python
umap_df = pd.DataFrame(index=adata.obs.index)
umap_df['umap0'] = adata.obsm['X_umap'][:,0]
umap_df['umap1'] = adata.obsm['X_umap'][:,1]

select = ['time_num', 'traj_score', 'clusters_v7', 'coarse_v7']
for s in select:
    umap_df[s] = adata.obs[s]
```

```python
scaler = MinMaxScaler()
umap_df['norm_traj'] = np.ravel(scaler.fit_transform(umap_df['traj_score'].values.reshape(-1,1)))
```
