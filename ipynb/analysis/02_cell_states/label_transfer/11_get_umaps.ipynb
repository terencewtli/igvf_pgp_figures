```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import scanpy as sc
import anndata
from sklearn.preprocessing import MinMaxScaler
```

```python
sc.set_figure_params(figsize=(5,5), frameon=False)
sc.settings.verbosity = 3
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna'
```

```python
adata = sc.read_h5ad(f'{projdir}/h5ad/processed/rna_finegrain.h5ad')
```

```python
adata
```

```python
clusters_path = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna/csv/label_transfer/snmct_xgboost_v7.csv'
clusters = pd.read_csv(clusters_path, sep='\t', index_col=0)
clusters.shape
```

```python
traj_score_path = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna/csv/label_transfer/snmct_traj_score.csv'
traj_score = pd.read_csv(traj_score_path, sep='\t', index_col=0)
traj_score.shape
```

```python
traj_score.head()
```

```python
adata.obs['cluster'] = clusters.reindex(adata.obs.index)['pred_coarse']
adata.obs['traj_score'] = traj_score.reindex(adata.obs.index)['traj_score']
```

```python
adata.obs['traj_score'].isna().sum()
```

```python
umap_df = pd.DataFrame(index=adata.obs.index)
umap_df['umap0'] = adata.obsm['X_umap'][:,0]
umap_df['umap1'] = adata.obsm['X_umap'][:,1]

select = ['time', 'traj_score', 'cluster']
for s in select:
    umap_df[s] = adata.obs[s]
```

```python
scaler = MinMaxScaler()
umap_df['norm_traj'] = np.ravel(scaler.fit_transform(umap_df['traj_score'].values.reshape(-1,1)))
```

```python
umap_df.head()
```

```python
indir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure1'
umap_df.to_csv(f'{indir}/snmct_rna_umap.csv', sep='\t')
```
