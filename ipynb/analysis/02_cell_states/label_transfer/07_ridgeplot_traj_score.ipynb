```r
library(ggplot2)
library(ggridges)
library(cowplot)
```

```r
projdir <- '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna/'
```

```r
multiome <- read.csv(paste0(projdir, '/csv/label_transfer/multiome_traj_score.csv'), sep='\t')
snmct <- read.csv(paste0(projdir, '/csv/label_transfer/snmct_traj_score.csv'), sep='\t')
```

```r
dim(multiome)
dim(snmct)
```

```r
head(multiome)
head(snmct)
```

```r
table(snmct$cluster)
```

```r
cat_order = rev(c('Start', 'Sendai', 'Skin', 'Inflamed', 'Delayed', 'Fail1', 'Inter1', 'Inter2', 'Fail2', 'IPS'))
```

```r
table(multiome$cluster)
```

```r
multiome$cluster <- factor(multiome$cluster, levels = cat_order)
snmct$cluster <- factor(snmct$cluster, levels = cat_order)
```

```r
p1 <- ggplot(multiome, aes(x = traj_score, y = cluster)) +
  geom_density_ridges(scale = 1, alpha = 0.7) +
  scale_x_continuous(limits = c(-0.5, 0.5)) +   # Set x-axis limits
  scale_fill_viridis_d(option = "C") +          # Optional: change fill color
  labs(x = "Trajectory Score", y = "", title = 'Multiome') +
  theme_minimal() +
  theme_ridges()                                # Add ridge plot styling

p1
```

```r
p2 <- ggplot(snmct, aes(x = traj_score, y = cluster)) +
  geom_density_ridges(scale = 1, alpha = 0.7) +
  scale_x_continuous(limits = c(-0.5, 0.5)) +   # Set x-axis limits
  scale_fill_viridis_d(option = "C") +          # Optional: change fill color
  labs(x = "Trajectory Score", y = "", title = 'snmCT') +
  theme_minimal() +
  theme_ridges()                                # Add ridge plot styling

p2
```

```r
combined_plot <- plot_grid(p1, p2,
                           ncol = 2,
                           align = "v",
                           labels = c("A", "B"),
                           label_size = 12)
title <- ggdraw() + draw_label("Per-cluster trajectory score distributions", fontface = 'bold', size = 14)

final_plot <- plot_grid(title, combined_plot, ncol = 1, rel_heights = c(0.1, 1))

options(repr.plot.width=10, repr.plot.height=10)

print(final_plot)
```
