```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import scanpy as sc
import anndata

from scipy.io import mmread, mmwrite
from scipy.sparse import csr_matrix

from sklearn.preprocessing import binarize
```

```python
sc.set_figure_params(figsize=(5,5), frameon=False)
sc.settings.verbosity = 3
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna'
```

```python
mtx = mmread(f'{projdir}/mtx/old/all_raw.mtx')
```

```python
mtx
```

```python
genes = pd.read_csv(f'{projdir}/csv/rna/all_rna_genes.csv', sep='\t', header=None, index_col=0)
nuclei = pd.read_csv(f'{projdir}/csv/rna/all_rna_nuclei.csv', sep='\t', header=None, index_col=0)
genes.shape, nuclei.shape
```

```python
adata = anndata.AnnData(X=csr_matrix(mtx), obs=genes, var=nuclei).T
```

```python
adata
```

```python
adata.var.index.name = 'ensg'
adata.obs.index.name = 'cell'
```

```python
adata.write_h5ad(f'{projdir}/h5ad/raw/snmct.h5ad')
```
