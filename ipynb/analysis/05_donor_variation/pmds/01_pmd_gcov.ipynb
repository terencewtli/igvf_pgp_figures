```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob

import itertools
from pybedtools import BedTool
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/pseudobulk_allc'
clusts = list(np.loadtxt(f'{projdir}/txt/allc_table/rna_cluster_donor.txt', dtype=str))
```

```python
indir = f'{projdir}/pmds/dnmtools/rna_cluster_donor'
pmds = pd.read_csv(f'{indir}/merged/start_merged.filt_mcg.sorted.bed.gz', sep='\t', header=None)
pmds.index = [f'{x}_{y}' for x,y in zip(pmds[0], pmds[1])]
pmds.shape
```

```python
%%time

genome = '/u/project/cluo/terencew/reference/hg38_igvf/GRCh38.autosomal.chromsizes'

pmds_bt = BedTool.from_dataframe(pmds)
gcov = pmds_bt.genome_coverage(g=genome)
gcov_df = gcov.to_dataframe()
gcov_df.shape
```

```python
gcov_df.head()
```

```python
gcov_df[gcov_df['chrom'] == 'genome']
```

```python
pmd_cov = gcov_df[gcov_df['start'] == 1].iloc[:-1,:]

fig, axes = plt.subplots(1, figsize=(6, 5))

x_lab = 'chrom'
y_lab = 'score'
hue = 'type'

ax = sns.barplot(data=pmd_cov, x=x_lab, y=y_lab)
                 
ax.set_xlabel('Chromosome', fontsize=8)
ax.set_ylabel('PMD coverage', fontsize=8)
ax.set_title('', fontsize=20)
ax.grid(False)
ax.set_ylim(0, 0.5)

ax.tick_params(axis='x', labelrotation=45)
```

```python
outdir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure3'
gcov_df.to_csv(f'{outdir}/pmd_union.cov.csv', sep='\t')
```
