```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT'
donors = list(np.loadtxt(f'{projdir}/mc/txt/donors.txt', dtype=str))
```

```python
meta = pd.read_csv(f'{projdir}/mc/csv/meta/passqc_cells.csv.gz', sep='\t', header=0, index_col=0)
meta.shape
```

```python
meta.head()
```

```python
meta['donor_time'] = [f'{x}_{y}' for x,y in zip(meta['line'], meta['time'])]
```

```python
meta.sort_values(by=['time', 'line'], inplace=True)
```

```python
meta['time_num'] = meta['time']
```

```python
meta['time'] = [str(x) for x in meta['time']]
```

```python
hexes = ['#1d8348', '#5b2c6f', '#922b21', '#f1c40f']
```

```python
fig, axes = plt.subplots(1, figsize=(4, 6))

x_lab = 'time'
y_lab = 'DNA_mCGfrac'
hue = 'line'

ax = sns.boxplot(data=meta, x=x_lab, y=y_lab, hue=hue,
                 showfliers=False, palette=hexes)
ax.set_xlabel('Time', fontsize=16)
ax.set_ylabel('mCG', fontsize=16)
ax.set_title('', fontsize=20)

ax.tick_params(axis='x', labelsize=14, labelrotation=45)
sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
```

### by cluster

```python
clust_path = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna/csv/label_transfer/snmct_xgboost_v7.csv'
clusters = pd.read_csv(clust_path, sep='\t', index_col=0)
clusters.shape
```

```python
clusters.head()
```

```python
meta['cluster'] = clusters.reindex(meta.index)['pred_coarse']
final_meta = meta[~meta['cluster'].isna()]
```

```python
fig, axes = plt.subplots(1, figsize=(4, 6))

x_lab = 'cluster'
y_lab = 'DNA_mCGfrac'
hue = 'line'

ax = sns.boxplot(data=final_meta, x=x_lab, y=y_lab, hue=hue,
                 showfliers=False, palette=hexes)
ax.set_xlabel('Time', fontsize=16)
ax.set_ylabel('mCG', fontsize=16)
ax.set_title('', fontsize=20)

ax.tick_params(axis='x', labelsize=14, labelrotation=45)
sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
```

```python
outdir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv/figure2'
select = ['line', 'cluster', 'DNA_mCGfrac']
final_meta[select].to_csv(f'{outdir}/cluster_donor_mcg.csv', sep='\t')
```

### rev class

```python
rev_class_path = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/rna/csv/label_transfer/CL1.metadata.with.ys3_v7_revclass.txt'
rev_class = pd.read_csv(rev_class_path, sep='\t', index_col=0)
rev_class.shape
```
