```r
clust <- 'IPS'
tmp.path.table <- full.path.table[full.path.table$Path==clust,]
tmp.path.table$Node_Cluster <- paste0(sapply(strsplit(tmp.path.table$Node,"_"),"[[",2),"_",sapply(strsplit(tmp.path.table$Node,"_"),"[[",3) )
ggplot(donor.metadata)+
  geom_point(data=markers, aes(X,Y), color="grey85")+
  geom_point(data=markers[markers$Cluster %in% tmp.path.table$Node_Cluster,], aes(X,Y, color=Cluster))+
  geom_point(data=markers, aes(X,Y), color="white", alpha=0.05)+
  # scale_colour_discrete(type =palette.vector)+
  geom_line(data=full.path.table, aes(meanX, meanY, group=Group), linewidth=0.3)+
  geom_line(data=full.path.table[full.path.table$Path==clust,],aes(meanX, meanY, group=Group2), color="red", linewidth=2)+
  geom_point(aes(meanX,meanY), color="black", size=3)+
  geom_point(aes(meanX,meanY), color="white", size=1)+
  geom_label(data=donor.metadata[!duplicated(donor.metadata$Cluster) & donor.metadata$Cluster %in% tmp.path.table$Node_Cluster,],aes(meanX,meanY, label=Cluster),vjust=(-2))+
  labs( x="UMAP 1", y="UMAP 2", color="Clusters")+
  theme_classic()+
  theme(legend.position = "")
```
