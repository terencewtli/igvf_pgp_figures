```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from concurrent.futures import ProcessPoolExecutor, as_completed
from tqdm import tqdm
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/pseudobulk_allc'
tmp_order = ['Start', 'Sendai', 'Delayed', 'Fail1', 'Inter1',  'Inter2', 'Fail2','IPS']
```

```python
# dmrs.shape
```

```python
indir = f'{projdir}/csv/dmrs/rna_cluster/by_group'

def read_dmrs(s):
    hypo_dmrs = pd.read_csv(f'{indir}/{s}_hypo.csv', sep='\t', header=None)
    hyper_dmrs = pd.read_csv(f'{indir}/{s}_hyper.csv', sep='\t', header=None)
    return hypo_dmrs, hyper_dmrs

with ProcessPoolExecutor() as executor:
    results = list(tqdm(executor.map(read_dmrs, tmp_order), total=len(tmp_order)))
```

```python
hypo = [x[0] for x in results]
hyper = [x[1] for x in results]
n_hypo = [x.shape[0] for x in hypo]
n_hyper = [x.shape[0] for x in hyper]
```

```python
dmr_summary = pd.DataFrame(zip(n_hypo, n_hyper))
dmr_summary.columns = ['hypomethylated', 'hypermethylated']
dmr_summary['cluster'] = tmp_order
dmr_summary.shape
```

```python
dmr_summary.head()
```

```python
to_plot = dmr_summary.melt(id_vars=['cluster'], value_vars=['hypomethylated', 'hypermethylated'])
to_plot.head()
```

```python
fig, axes = plt.subplots(1, figsize=(6, 6))

x_lab = 'cluster'
y_lab = 'value'
hue = 'variable'

ax = sns.barplot(data=to_plot, x=x_lab, y=y_lab, hue=hue,
                 order=tmp_order)
ax.set_xlabel('Cluster', fontsize=16)
ax.set_ylabel('# DMRs', fontsize=16)
ax.set_title('', fontsize=20)
ax.grid(False)

ax.tick_params(axis='x', labelsize=14, labelrotation=45)
sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
```

```python
indir = '/u/project/cluo/terencew/igvf/2023_YR2/final_figures/csv'
to_plot.to_csv(f'{indir}/figure1/rna_cluster_num_dmrs.csv', sep='\t')
```
